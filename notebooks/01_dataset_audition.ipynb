## Dataset comparison and selection

The three candidate datasets differ in size, analytical opportunities, and data-quality risks. The NSW Train Occupancy dataset contains 50,250 observations and 14 variables, with relatively low missingness and approximately 0.5% exact duplicate rows. The Air Travel Delay dataset contains 101,000 observations and 29 variables, providing substantial opportunities to investigate delay patterns, but several delay-related variables have high levels of missingness. The Hotel Booking dataset contains 119,987 observations and 32 variables and offers rich information about booking behaviour, but approximately 26.1% of its rows are exact duplicates, creating a substantial data-quality concern.

The NSW Train Occupancy dataset is selected for further analysis because it provides a relatively clean and manageable evidence base while directly supporting the assignment mission of advising a transport operations team about crowding patterns and opportunities for improved service planning. Its variables describe train service characteristics, station information, timing, occupancy status and occupancy range, providing opportunities to investigate when and where crowding occurs.
